Decorators can be thought of as functions which modify the *functionality* of another function. They help to make your code shorter and more "Pythonic".

````python
s = 'Global Variable'

def check_for_locals():
    print(locals())
````

## Functions within functions
Great! So we've seen how we can treat functions as objects, now let's see how we can define functions inside of other functions:

````python
def hello(name='Jose'):
    print('The hello() function has been executed')
    
    def greet():
        return '\t This is inside the greet() function'
    
    def welcome():
        return "\t This is inside the welcome() function"
    
    print(greet())
    print(welcome())
    print("Now we are back inside the hello() function")
````

````python
hello()
````

````python
welcome()
````

Note how due to scope, the welcome() function is not defined outside of the hello() function. Now lets learn about returning functions from within functions:
## Returning Functions

````python
def hello(name='Jose'):
    
    def greet():
        return '\t This is inside the greet() function'
    
    def welcome():
        return "\t This is inside the welcome() function"
    
    if name == 'Jose':
        return greet
    else:
        return welcome
````

Now let's see what function is returned if we set x = hello(), note how the empty parentheses means that name has been defined as Jose.

````python
x = hello()
````

````python
x
````

Great! Now we can see how x is pointing to the greet function inside of the hello function.

````python
print(x())
````

Let's take a quick look at the code again. 

In the <code>if</code>/<code>else</code> clause we are returning <code>greet</code> and <code>welcome</code>, not <code>greet()</code> and <code>welcome()</code>. 

This is because when you put a pair of parentheses after it, the function gets executed; whereas if you don’t put parentheses after it, then it can be passed around and can be assigned to other variables without executing it.

When we write <code>x = hello()</code>, hello() gets executed and because the name is Jose by default, the function <code>greet</code> is returned. If we change the statement to <code>x = hello(name = "Sam")</code> then the <code>welcome</code> function will be returned. We can also do <code>print(hello()())</code> which outputs *This is inside the greet() function*.

## Functions as Arguments
Now let's see how we can pass functions as arguments into other functions:

````python
def hello():
    return 'Hi Jose!'

def other(func):
    print('Other code would go here')
    print(func())
````

````python
other(hello)
````

Great! Note how we can pass the functions as objects and then use them within other functions. Now we can get started with writing our first decorator:

## Creating a Decorator
In the previous example we actually manually created a Decorator. Here we will modify it to make its use case clear:

````python
def new_decorator(func):

    def wrap_func():
        print("Code would be here, before executing the func")

        func()

        print("Code here will execute after the func()")

    return wrap_func

def func_needs_decorator():
    print("This function is in need of a Decorator")
````

````python
func_needs_decorator()
````

````python
# Reassign func_needs_decorator
func_needs_decorator = new_decorator(func_needs_decorator)
````

````python
func_needs_decorator()
````

So what just happened here? A decorator simply wrapped the function and modified its behavior. Now let's understand how we can rewrite this code using the @ symbol, which is what Python uses for Decorators:

````python
@new_decorator
def func_needs_decorator():
    print("This function is in need of a Decorator")
````

````python
func_needs_decorator()
````

**Great! You've now built a Decorator manually and then saw how we can use the @ symbol in Python to automate this and clean our code. You'll run into Decorators a lot if you begin using Python for Web Development, such as Flask or Django!**

# Azure Functions Decorators

## Introduzione ai Decoratori in Azure Functions

Azure Functions è un servizio serverless di Microsoft Azure che permette di eseguire codice in risposta a eventi senza doversi preoccupare dell'infrastruttura. A partire dalla versione Python V2 programming model, Azure Functions utilizza i decoratori in modo estensivo per definire le funzioni e i loro trigger.

I decoratori di Azure Functions sono un esempio pratico di come i decoratori Python vengono utilizzati in applicazioni reali di livello enterprise. Essi forniscono un modo elegante e "Pythonic" per configurare le funzioni serverless.

## Come Funzionano i Decoratori di Azure Functions

I decoratori di Azure Functions servono principalmente a:
1. **Definire i trigger** - Specificare quale evento attiva la funzione (HTTP request, timer, queue message, ecc.)
2. **Configurare i binding** - Definire input e output della funzione
3. **Aggiungere metadati** - Specificare configurazioni come metodi HTTP, percorsi, autenticazione

## Esempio Base: HTTP Trigger

Vediamo un esempio semplice di una Azure Function con un trigger HTTP:

````python
import azure.functions as func
import logging

# Creazione dell'app Azure Functions
app = func.FunctionApp()

# Decoratore che definisce un trigger HTTP
@app.function_name(name="HttpExample")
@app.route(route="hello", methods=["GET", "POST"])
def main(req: func.HttpRequest) -> func.HttpResponse:
    """
    Questa funzione risponde a richieste HTTP GET e POST sul percorso '/hello'
    """
    logging.info('Python HTTP trigger function processed a request.')

    name = req.params.get('name')
    if not name:
        try:
            req_body = req.get_json()
        except ValueError:
            pass
        else:
            name = req_body.get('name')

    if name:
        return func.HttpResponse(f"Hello, {name}! This HTTP triggered function executed successfully.")
    else:
        return func.HttpResponse(
             "This HTTP triggered function executed successfully. Pass a name in the query string or in the request body for a personalized response.",
             status_code=200
        )
````

### Analisi dei Decoratori

Nell'esempio sopra abbiamo utilizzato due decoratori:

1. **`@app.function_name(name="HttpExample")`** - Definisce il nome della funzione in Azure
2. **`@app.route(route="hello", methods=["GET", "POST"])`** - Definisce:
   - Il percorso URL (`/hello`)
   - I metodi HTTP consentiti (`GET` e `POST`)

Questi decoratori "avvolgono" la funzione `main()` aggiungendo metadati e funzionalità senza modificare la logica interna della funzione.

## Tipi di Trigger Comuni

Azure Functions supporta vari tipi di trigger attraverso decoratori specifici:

### 1. Timer Trigger
Esegue la funzione su una pianificazione (come un cron job):

````python
@app.function_name(name="TimerTriggerExample")
@app.schedule(schedule="0 */5 * * * *", arg_name="mytimer", run_on_startup=False)
def timer_trigger(mytimer: func.TimerRequest) -> None:
    """
    Questa funzione viene eseguita ogni 5 minuti
    La sintassi dello schedule è: secondi minuti ore giorni mesi giorni-settimana
    """
    logging.info('Timer trigger function executed.')
    if mytimer.past_due:
        logging.info('The timer is past due!')
    
    # Logica della funzione
    logging.info('Processing scheduled task...')
````

### 2. Queue Trigger
Si attiva quando viene aggiunto un messaggio a una coda di Azure Storage:

````python
@app.function_name(name="QueueTriggerExample")
@app.queue_trigger(arg_name="msg", queue_name="myqueue", connection="AzureWebJobsStorage")
def queue_trigger(msg: func.QueueMessage) -> None:
    """
    Questa funzione processa messaggi dalla coda 'myqueue'
    """
    logging.info(f'Python queue trigger function processed a queue item: {msg.get_body().decode("utf-8")}')
    
    # Elaborazione del messaggio
    message_content = msg.get_body().decode('utf-8')
    logging.info(f'Processing message: {message_content}')
````

### 3. Blob Trigger
Si attiva quando viene aggiunto o modificato un blob in Azure Blob Storage:

````python
@app.function_name(name="BlobTriggerExample")
@app.blob_trigger(arg_name="myblob", path="mycontainer/{name}", connection="AzureWebJobsStorage")
def blob_trigger(myblob: func.InputStream) -> None:
    """
    Questa funzione viene attivata quando viene caricato un file nel container 'mycontainer'
    """
    logging.info(f"Python blob trigger function processed blob \n"
                f"Name: {myblob.name}\n"
                f"Blob Size: {myblob.length} bytes")
    
    # Lettura del contenuto del blob
    blob_content = myblob.read()
    logging.info(f"Processing blob content...")
````

## Binding di Input e Output

Azure Functions utilizza anche decoratori per definire binding di input e output, permettendo alla funzione di leggere e scrivere su vari servizi Azure senza codice boilerplate:

````python
@app.function_name(name="HttpTriggerWithBindings")
@app.route(route="process", methods=["POST"])
@app.blob_input(arg_name="inputblob", path="input/{name}", connection="AzureWebJobsStorage")
@app.blob_output(arg_name="outputblob", path="output/{name}", connection="AzureWebJobsStorage")
@app.queue_output(arg_name="outputqueue", queue_name="outqueue", connection="AzureWebJobsStorage")
def process_with_bindings(
    req: func.HttpRequest,
    inputblob: func.InputStream,
    outputblob: func.Out[bytes],
    outputqueue: func.Out[str]
) -> func.HttpResponse:
    """
    Esempio complesso con multiple bindings:
    - Trigger HTTP per avviare il processo
    - Input binding per leggere da blob storage
    - Output binding per scrivere su blob storage
    - Output binding per scrivere su una coda
    """
    logging.info('Processing request with multiple bindings')
    
    # Legge dall'input blob
    input_content = inputblob.read()
    
    # Processa i dati
    processed_data = input_content.upper()  # Esempio semplice
    
    # Scrive sull'output blob
    outputblob.set(processed_data)
    
    # Invia un messaggio alla coda
    outputqueue.set("Processing completed successfully")
    
    return func.HttpResponse(
        "Data processed and outputs written successfully!",
        status_code=200
    )
````

## Come Funzionano Internamente i Decoratori di Azure Functions

I decoratori di Azure Functions sono un'implementazione sofisticata del pattern decorator che abbiamo visto in precedenza. Vediamo cosa succede "dietro le quinte":

### 1. Registrazione della Funzione
Quando usi `@app.function_name()`, il decoratore:
- Registra la funzione nell'oggetto `FunctionApp`
- Crea metadati per Azure
- Associa la funzione a un nome univoco

### 2. Configurazione dei Trigger
I decoratori di trigger come `@app.route()` o `@app.schedule()`:
- Definiscono quando la funzione deve essere eseguita
- Configurano i parametri di input
- Creano le regole di binding

### 3. Binding dei Dati
I decoratori di binding:
- Gestiscono automaticamente la connessione ai servizi Azure
- Serializzano/deserializzano i dati
- Gestiscono l'autenticazione e le connessioni

## Esempio Simulato: Come Funziona un Decoratore di Azure Functions

Vediamo un esempio semplificato di come potrebbe essere implementato internamente un decoratore simile a quelli di Azure Functions:

````python
# Simulazione semplificata di come funziona un decoratore di Azure Functions

class SimpleFunctionApp:
    """
    Classe semplificata che simula il comportamento di azure.functions.FunctionApp
    """
    def __init__(self):
        self.functions = {}  # Dizionario per memorizzare le funzioni registrate
    
    def function_name(self, name):
        """
        Decoratore che registra una funzione con un nome
        """
        def decorator(func):
            # Registra la funzione
            self.functions[name] = {
                'function': func,
                'name': name,
                'metadata': {}
            }
            print(f"Funzione '{name}' registrata")
            return func
        return decorator
    
    def route(self, route, methods=None):
        """
        Decoratore che aggiunge metadati di routing HTTP
        """
        if methods is None:
            methods = ["GET"]
            
        def decorator(func):
            # Trova la funzione registrata e aggiungi metadati
            for name, func_data in self.functions.items():
                if func_data['function'] == func:
                    func_data['metadata']['route'] = route
                    func_data['metadata']['methods'] = methods
                    print(f"Route '{route}' configurata con metodi {methods}")
            
            # Wrapper che aggiunge funzionalità HTTP
            def wrapper(*args, **kwargs):
                print(f"Richiesta HTTP ricevuta su '{route}'")
                result = func(*args, **kwargs)
                print(f"Risposta HTTP inviata")
                return result
            
            return wrapper
        return decorator
    
    def list_functions(self):
        """Mostra tutte le funzioni registrate"""
        print("\n=== Funzioni Registrate ===")
        for name, data in self.functions.items():
            print(f"\nNome: {name}")
            print(f"Metadata: {data['metadata']}")

# Creiamo un'istanza della nostra app simulata
simple_app = SimpleFunctionApp()

# Utilizziamo i decoratori come in Azure Functions
@simple_app.function_name(name="HelloFunction")
@simple_app.route(route="hello", methods=["GET", "POST"])
def hello_example(name="World"):
    return f"Hello, {name}!"

# Testiamo la funzione
print("\n=== Test della funzione ===")
result = hello_example("Azure")
print(f"Risultato: {result}")

# Mostriamo i dettagli delle funzioni registrate
simple_app.list_functions()
````

## Esempio Pratico Completo: API REST con Azure Functions

Ecco un esempio completo di un'API REST che utilizza vari decoratori di Azure Functions:

````python
import azure.functions as func
import logging
import json

# Inizializzazione dell'app
app = func.FunctionApp()

# ===== CRUD API per Prodotti =====

# CREATE - Aggiunge un nuovo prodotto
@app.function_name(name="CreateProduct")
@app.route(route="products", methods=["POST"], auth_level=func.AuthLevel.FUNCTION)
@app.cosmos_db_output(
    arg_name="outputDocument",
    database_name="ProductsDB",
    container_name="Products",
    connection="CosmosDBConnection"
)
def create_product(req: func.HttpRequest, outputDocument: func.Out[func.Document]) -> func.HttpResponse:
    """
    Crea un nuovo prodotto nel database
    POST /api/products
    Body: {"id": "1", "name": "Product Name", "price": 99.99}
    """
    try:
        product = req.get_json()
        
        # Validazione base
        if not product.get('id') or not product.get('name'):
            return func.HttpResponse(
                "Missing required fields: id and name",
                status_code=400
            )
        
        # Salva nel database tramite output binding
        outputDocument.set(func.Document.from_dict(product))
        
        return func.HttpResponse(
            json.dumps({"message": "Product created", "product": product}),
            mimetype="application/json",
            status_code=201
        )
    except ValueError:
        return func.HttpResponse("Invalid JSON", status_code=400)

# READ - Ottiene un prodotto specifico
@app.function_name(name="GetProduct")
@app.route(route="products/{id}", methods=["GET"])
@app.cosmos_db_input(
    arg_name="inputDocument",
    database_name="ProductsDB",
    container_name="Products",
    id="{id}",
    partition_key="{id}",
    connection="CosmosDBConnection"
)
def get_product(req: func.HttpRequest, inputDocument: func.DocumentList) -> func.HttpResponse:
    """
    Ottiene un prodotto per ID
    GET /api/products/{id}
    """
    if not inputDocument:
        return func.HttpResponse("Product not found", status_code=404)
    
    product = inputDocument[0].to_dict()
    return func.HttpResponse(
        json.dumps(product),
        mimetype="application/json"
    )

# LIST - Ottiene tutti i prodotti
@app.function_name(name="ListProducts")
@app.route(route="products", methods=["GET"])
@app.cosmos_db_input(
    arg_name="inputDocuments",
    database_name="ProductsDB",
    container_name="Products",
    connection="CosmosDBConnection"
)
def list_products(req: func.HttpRequest, inputDocuments: func.DocumentList) -> func.HttpResponse:
    """
    Ottiene lista di tutti i prodotti
    GET /api/products
    """
    products = [doc.to_dict() for doc in inputDocuments]
    return func.HttpResponse(
        json.dumps({"products": products, "count": len(products)}),
        mimetype="application/json"
    )

# UPDATE - Aggiorna un prodotto
@app.function_name(name="UpdateProduct")
@app.route(route="products/{id}", methods=["PUT"])
@app.cosmos_db_input(
    arg_name="inputDocument",
    database_name="ProductsDB",
    container_name="Products",
    id="{id}",
    partition_key="{id}",
    connection="CosmosDBConnection"
)
@app.cosmos_db_output(
    arg_name="outputDocument",
    database_name="ProductsDB",
    container_name="Products",
    connection="CosmosDBConnection"
)
def update_product(
    req: func.HttpRequest,
    inputDocument: func.DocumentList,
    outputDocument: func.Out[func.Document]
) -> func.HttpResponse:
    """
    Aggiorna un prodotto esistente
    PUT /api/products/{id}
    Body: {"name": "New Name", "price": 149.99}
    """
    if not inputDocument:
        return func.HttpResponse("Product not found", status_code=404)
    
    try:
        updates = req.get_json()
        product = inputDocument[0].to_dict()
        
        # Aggiorna i campi
        product.update(updates)
        
        # Salva il prodotto aggiornato
        outputDocument.set(func.Document.from_dict(product))
        
        return func.HttpResponse(
            json.dumps({"message": "Product updated", "product": product}),
            mimetype="application/json"
        )
    except ValueError:
        return func.HttpResponse("Invalid JSON", status_code=400)

# DELETE - Elimina un prodotto (con notifica via queue)
@app.function_name(name="DeleteProduct")
@app.route(route="products/{id}", methods=["DELETE"])
@app.queue_output(
    arg_name="notificationQueue",
    queue_name="product-deletions",
    connection="AzureWebJobsStorage"
)
def delete_product(
    req: func.HttpRequest,
    notificationQueue: func.Out[str]
) -> func.HttpResponse:
    """
    Elimina un prodotto (implementazione semplificata)
    DELETE /api/products/{id}
    
    Invia una notifica alla coda per processing asincrono
    """
    product_id = req.route_params.get('id')
    
    # In un caso reale, elimineresti dal database qui
    # Per semplicità, inviamo solo la notifica
    
    notification = json.dumps({
        "action": "delete",
        "product_id": product_id,
        "timestamp": "2025-01-01T00:00:00Z"
    })
    
    notificationQueue.set(notification)
    
    return func.HttpResponse(
        json.dumps({"message": f"Product {product_id} deletion queued"}),
        mimetype="application/json"
    )

# Funzione di background che processa le eliminazioni
@app.function_name(name="ProcessDeletions")
@app.queue_trigger(
    arg_name="msg",
    queue_name="product-deletions",
    connection="AzureWebJobsStorage"
)
def process_deletions(msg: func.QueueMessage) -> None:
    """
    Processa le eliminazioni in modo asincrono
    Trigger: Queue message
    """
    message_data = json.loads(msg.get_body().decode('utf-8'))
    product_id = message_data.get('product_id')
    
    logging.info(f"Processing deletion for product: {product_id}")
    
    # Qui andresti effettivamente ad eliminare dal database
    # e magari inviare email di notifica, aggiornare analytics, ecc.
    
    logging.info(f"Product {product_id} successfully deleted")
````

## Best Practices con Decoratori Azure Functions

### 1. Ordine dei Decoratori
L'ordine è importante! Di solito:
```python
@app.function_name(...)      # Primo: definisce la funzione
@app.route(...)              # Secondo: definisce il trigger

    pass
```

### 2. Naming Conventions
- Usa nomi descrittivi per `function_name`
- I nomi delle route dovrebbero seguire convenzioni RESTful
- I parametri `arg_name` dovrebbero essere descrittivi

### 3. Error Handling
Gestisci sempre gli errori nelle funzioni:
```python
try:
    # Logica della funzione
    pass

    return func.HttpResponse("Internal error", status_code=500)
```

### 4. Logging
Usa il logging per debugging e monitoring:
```python
logging.info('Processing request...')
logging.error(f'Error occurred: {error_message}')
```

## Confronto: Decoratori Standard vs Azure Functions

Per capire meglio la potenza dei decoratori di Azure Functions, confrontiamoli con i decoratori standard di Python:

### Decoratore Standard Python

````python
# Decoratore Standard - Aggiunge logging
def log_execution(func):
    """Decoratore standard che logga l'esecuzione di una funzione"""
    def wrapper(*args, **kwargs):
        print(f"Executing {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Finished {func.__name__}")
        return result
    return wrapper

@log_execution
def add_numbers(a, b):
    """Semplice funzione che somma due numeri"""
    return a + b

# Test
result = add_numbers(5, 3)
print(f"Result: {result}")
````

### Decoratore Azure Functions - Aggiunge Trigger e Infrastruttura

I decoratori di Azure Functions fanno molto di più:
- Non solo avvolgono la funzione
- Registrano metadati per l'infrastruttura cloud
- Gestiscono connessioni, autenticazione, serializzazione
- Integrano con servizi Azure (Storage, Cosmos DB, Service Bus, ecc.)

```python
# Lo stesso decoratore in stile Azure Functions farebbe:
@app.function_name(name="AddNumbers")
@app.route(route="add", methods=["POST"])
@app.cosmos_db_output(...)  # Salva il risultato in un database
@app.queue_output(...)      # Invia notifica a una coda
def add_numbers(req: func.HttpRequest, ...) -> func.HttpResponse:
    # La funzione ora è accessibile via HTTP
    # Può leggere/scrivere su database
    # Può inviare messaggi a code
    # Tutto gestito automaticamente dai decoratori
    pass
```

**Differenze Chiave:**
1. **Scopo**: Standard = modifica comportamento; Azure = configura infrastruttura + modifica comportamento
2. **Complessità**: Standard = semplici wrapper; Azure = gestione completa di servizi cloud
3. **Metadata**: Standard = nessuno; Azure = estensivi metadata per runtime
4. **Composizione**: Azure permette stacking di multiple funzionalità tramite decoratori multipli

## Decoratori Custom per Azure Functions

````python
import time
from functools import wraps

# Decoratore custom per validare il JSON
def validate_json_body(required_fields):
    """
    Decoratore che valida che il body della richiesta contenga i campi richiesti
    """
    def decorator(func):
        @wraps(func)
        def wrapper(req: func.HttpRequest, *args, **kwargs):
            try:
                body = req.get_json()
                
                # Controlla che tutti i campi richiesti siano presenti
                missing_fields = [field for field in required_fields if field not in body]
                
                if missing_fields:
                    return func.HttpResponse(
                        f"Missing required fields: {', '.join(missing_fields)}",
                        status_code=400
                    )
                
                # Se la validazione passa, esegue la funzione
                return func(req, *args, **kwargs)
                
            except ValueError:
                return func.HttpResponse(
                    "Invalid JSON in request body",
                    status_code=400
                )
        return wrapper
    return decorator

# Decoratore custom per timing
def measure_execution_time(func):
    """
    Decoratore che misura il tempo di esecuzione della funzione
    """
    @wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        
        execution_time = end_time - start_time
        logging.info(f"Function {func.__name__} executed in {execution_time:.4f} seconds")
        
        return result
    return wrapper

# Decoratore custom per rate limiting (semplificato)
def rate_limit(max_calls_per_minute=60):
    """
    Decoratore semplificato per rate limiting
    """
    calls = {}
    
    def decorator(func):
        @wraps(func)
        def wrapper(req: func.HttpRequest, *args, **kwargs):
            # In produzione, useresti Redis o simile per tracking distribuito
            client_ip = req.headers.get('X-Forwarded-For', 'unknown')
            current_minute = int(time.time() / 60)
            
            key = f"{client_ip}:{current_minute}"
            
            if key in calls:
                calls[key] += 1
                if calls[key] > max_calls_per_minute:
                    return func.HttpResponse(
                        "Rate limit exceeded. Please try again later.",
                        status_code=429
                    )
            else:
                calls[key] = 1
            
            return func(req, *args, **kwargs)
        return wrapper
    return decorator

# Esempio di utilizzo combinato con decoratori Azure Functions
# (questo è pseudo-codice per illustrare il concetto)
"""
@app.function_name(name="CreateUserWithValidation")
@app.route(route="users", methods=["POST"])
@rate_limit(max_calls_per_minute=10)
@validate_json_body(required_fields=['username', 'email', 'password'])
@measure_execution_time
def create_user(req: func.HttpRequest) -> func.HttpResponse:
    # Il body è già validato dai decoratori
    # Il rate limiting è già applicato
    # Il tempo di esecuzione viene automaticamente loggato
    
    user_data = req.get_json()
    
    # Logica per creare l'utente
    # ...
    
    return func.HttpResponse(
        json.dumps({"message": "User created successfully"}),
        mimetype="application/json",
        status_code=201
    )
"""

print("Decoratori custom definiti!")
print("Questi possono essere combinati con i decoratori di Azure Functions per aggiungere validazione,")
print("monitoring, rate limiting, e altre funzionalità cross-cutting.")
````

# Azure Functions Decorators

## Introduzione ai Decoratori in Azure Functions

Azure Functions è un servizio serverless di Microsoft Azure che permette di eseguire codice in risposta a eventi senza doversi preoccupare dell'infrastruttura. A partire dalla versione Python V2 programming model, Azure Functions utilizza i decoratori in modo estensivo per definire le funzioni e i loro trigger.

I decoratori di Azure Functions sono un esempio pratico di come i decoratori Python vengono utilizzati in applicazioni reali di livello enterprise. Essi forniscono un modo elegante e "Pythonic" per configurare le funzioni serverless.

## Come Funzionano i Decoratori di Azure Functions

I decoratori di Azure Functions servono principalmente a:
1. **Definire i trigger** - Specificare quale evento attiva la funzione (HTTP request, timer, queue message, ecc.)
2. **Configurare i binding** - Definire input e output della funzione
3. **Aggiungere metadati** - Specificare configurazioni come metodi HTTP, percorsi, autenticazione

## Esempio Base: HTTP Trigger

Vediamo un esempio semplice di una Azure Function con un trigger HTTP:

````python
import azure.functions as func
import logging

# Creazione dell'app Azure Functions
app = func.FunctionApp()

# Decoratore che definisce un trigger HTTP
@app.function_name(name="HttpExample")
@app.route(route="hello", methods=["GET", "POST"])
def main(req: func.HttpRequest) -> func.HttpResponse:
    """
    Questa funzione risponde a richieste HTTP GET e POST sul percorso '/hello'
    """
    logging.info('Python HTTP trigger function processed a request.')

    name = req.params.get('name')
    if not name:
        try:
            req_body = req.get_json()
        except ValueError:
            pass
        else:
            name = req_body.get('name')

    if name:
        return func.HttpResponse(f"Hello, {name}! This HTTP triggered function executed successfully.")
    else:
        return func.HttpResponse(
             "This HTTP triggered function executed successfully. Pass a name in the query string or in the request body for a personalized response.",
             status_code=200
        )
````

### Analisi dei Decoratori

Nell'esempio sopra abbiamo utilizzato due decoratori:

1. **`@app.function_name(name="HttpExample")`** - Definisce il nome della funzione in Azure
2. **`@app.route(route="hello", methods=["GET", "POST"])`** - Definisce:
   - Il percorso URL (`/hello`)
   - I metodi HTTP consentiti (`GET` e `POST`)

Questi decoratori "avvolgono" la funzione `main()` aggiungendo metadati e funzionalità senza modificare la logica interna della funzione.

## Tipi di Trigger Comuni

Azure Functions supporta vari tipi di trigger attraverso decoratori specifici:

### 1. Timer Trigger
Esegue la funzione su una pianificazione (come un cron job):

````python
@app.function_name(name="TimerTriggerExample")
@app.schedule(schedule="0 */5 * * * *", arg_name="mytimer", run_on_startup=False)
def timer_trigger(mytimer: func.TimerRequest) -> None:
    """
    Questa funzione viene eseguita ogni 5 minuti
    La sintassi dello schedule è: secondi minuti ore giorni mesi giorni-settimana
    """
    logging.info('Timer trigger function executed.')
    if mytimer.past_due:
        logging.info('The timer is past due!')
    
    # Logica della funzione
    logging.info('Processing scheduled task...')
````

### 2. Queue Trigger
Si attiva quando viene aggiunto un messaggio a una coda di Azure Storage:

````python
@app.function_name(name="QueueTriggerExample")
@app.queue_trigger(arg_name="msg", queue_name="myqueue", connection="AzureWebJobsStorage")
def queue_trigger(msg: func.QueueMessage) -> None:
    """
    Questa funzione processa messaggi dalla coda 'myqueue'
    """
    logging.info(f'Python queue trigger function processed a queue item: {msg.get_body().decode("utf-8")}')
    
    # Elaborazione del messaggio
    message_content = msg.get_body().decode('utf-8')
    logging.info(f'Processing message: {message_content}')
````

### 3. Blob Trigger
Si attiva quando viene aggiunto o modificato un blob in Azure Blob Storage:

````python
@app.function_name(name="BlobTriggerExample")
@app.blob_trigger(arg_name="myblob", path="mycontainer/{name}", connection="AzureWebJobsStorage")
def blob_trigger(myblob: func.InputStream) -> None:
    """
    Questa funzione viene attivata quando viene caricato un file nel container 'mycontainer'
    """
    logging.info(f"Python blob trigger function processed blob \n"
                f"Name: {myblob.name}\n"
                f"Blob Size: {myblob.length} bytes")
    
    # Lettura del contenuto del blob
    blob_content = myblob.read()
    logging.info(f"Processing blob content...")
````

## Binding di Input e Output

Azure Functions utilizza anche decoratori per definire binding di input e output, permettendo alla funzione di leggere e scrivere su vari servizi Azure senza codice boilerplate:

````python
@app.function_name(name="HttpTriggerWithBindings")
@app.route(route="process", methods=["POST"])
@app.blob_input(arg_name="inputblob", path="input/{name}", connection="AzureWebJobsStorage")
@app.blob_output(arg_name="outputblob", path="output/{name}", connection="AzureWebJobsStorage")
@app.queue_output(arg_name="outputqueue", queue_name="outqueue", connection="AzureWebJobsStorage")
def process_with_bindings(
    req: func.HttpRequest,
    inputblob: func.InputStream,
    outputblob: func.Out[bytes],
    outputqueue: func.Out[str]
) -> func.HttpResponse:
    """
    Esempio complesso con multiple bindings:
    - Trigger HTTP per avviare il processo
    - Input binding per leggere da blob storage
    - Output binding per scrivere su blob storage
    - Output binding per scrivere su una coda
    """
    logging.info('Processing request with multiple bindings')
    
    # Legge dall'input blob
    input_content = inputblob.read()
    
    # Processa i dati
    processed_data = input_content.upper()  # Esempio semplice
    
    # Scrive sull'output blob
    outputblob.set(processed_data)
    
    # Invia un messaggio alla coda
    outputqueue.set("Processing completed successfully")
    
    return func.HttpResponse(
        "Data processed and outputs written successfully!",
        status_code=200
    )
````

## Esempio Simulato: Come Funziona un Decoratore di Azure Functions

Vediamo un esempio semplificato di come potrebbe essere implementato internamente un decoratore simile a quelli di Azure Functions:

````python
# Simulazione semplificata di come funziona un decoratore di Azure Functions

class SimpleFunctionApp:
    """
    Classe semplificata che simula il comportamento di azure.functions.FunctionApp
    """
    def __init__(self):
        self.functions = {}  # Dizionario per memorizzare le funzioni registrate
    
    def function_name(self, name):
        """
        Decoratore che registra una funzione con un nome
        """
        def decorator(func):
            # Registra la funzione
            self.functions[name] = {
                'function': func,
                'name': name,
                'metadata': {}
            }
            print(f"Funzione '{name}' registrata")
            return func
        return decorator
    
    def route(self, route, methods=None):
        """
        Decoratore che aggiunge metadati di routing HTTP
        """
        if methods is None:
            methods = ["GET"]
            
        def decorator(func):
            # Trova la funzione registrata e aggiungi metadati
            for name, func_data in self.functions.items():
                if func_data['function'] == func:
                    func_data['metadata']['route'] = route
                    func_data['metadata']['methods'] = methods
                    print(f"Route '{route}' configurata con metodi {methods}")
            
            # Wrapper che aggiunge funzionalità HTTP
            def wrapper(*args, **kwargs):
                print(f"Richiesta HTTP ricevuta su '{route}'")
                result = func(*args, **kwargs)
                print(f"Risposta HTTP inviata")
                return result
            
            return wrapper
        return decorator
    
    def list_functions(self):
        """Mostra tutte le funzioni registrate"""
        print("\n=== Funzioni Registrate ===")
        for name, data in self.functions.items():
            print(f"\nNome: {name}")
            print(f"Metadata: {data['metadata']}")

# Creiamo un'istanza della nostra app simulata
simple_app = SimpleFunctionApp()

# Utilizziamo i decoratori come in Azure Functions
@simple_app.function_name(name="HelloFunction")
@simple_app.route(route="hello", methods=["GET", "POST"])
def hello_example(name="World"):
    return f"Hello, {name}!"

# Testiamo la funzione
print("\n=== Test della funzione ===")
result = hello_example("Azure")
print(f"Risultato: {result}")

# Mostriamo i dettagli delle funzioni registrate
simple_app.list_functions()
````

## Esempio Pratico Completo: API REST con Azure Functions

Ecco un esempio completo di un'API REST che utilizza vari decoratori di Azure Functions:

````python
import azure.functions as func
import logging
import json

# Inizializzazione dell'app
app = func.FunctionApp()

# ===== CRUD API per Prodotti =====

# CREATE - Aggiunge un nuovo prodotto
@app.function_name(name="CreateProduct")
@app.route(route="products", methods=["POST"], auth_level=func.AuthLevel.FUNCTION)
@app.cosmos_db_output(
    arg_name="outputDocument",
    database_name="ProductsDB",
    container_name="Products",
    connection="CosmosDBConnection"
)
def create_product(req: func.HttpRequest, outputDocument: func.Out[func.Document]) -> func.HttpResponse:
    """
    Crea un nuovo prodotto nel database
    POST /api/products
    Body: {"id": "1", "name": "Product Name", "price": 99.99}
    """
    try:
        product = req.get_json()
        
        # Validazione base
        if not product.get('id') or not product.get('name'):
            return func.HttpResponse(
                "Missing required fields: id and name",
                status_code=400
            )
        
        # Salva nel database tramite output binding
        outputDocument.set(func.Document.from_dict(product))
        
        return func.HttpResponse(
            json.dumps({"message": "Product created", "product": product}),
            mimetype="application/json",
            status_code=201
        )
    except ValueError:
        return func.HttpResponse("Invalid JSON", status_code=400)

# READ - Ottiene un prodotto specifico
@app.function_name(name="GetProduct")
@app.route(route="products/{id}", methods=["GET"])
@app.cosmos_db_input(
    arg_name="inputDocument",
    database_name="ProductsDB",
    container_name="Products",
    id="{id}",
    partition_key="{id}",
    connection="CosmosDBConnection"
)
def get_product(req: func.HttpRequest, inputDocument: func.DocumentList) -> func.HttpResponse:
    """
    Ottiene un prodotto per ID
    GET /api/products/{id}
    """
    if not inputDocument:
        return func.HttpResponse("Product not found", status_code=404)
    
    product = inputDocument[0].to_dict()
    return func.HttpResponse(
        json.dumps(product),
        mimetype="application/json"
    )

# LIST - Ottiene tutti i prodotti
@app.function_name(name="ListProducts")
@app.route(route="products", methods=["GET"])
@app.cosmos_db_input(
    arg_name="inputDocuments",
    database_name="ProductsDB",
    container_name="Products",
    connection="CosmosDBConnection"
)
def list_products(req: func.HttpRequest, inputDocuments: func.DocumentList) -> func.HttpResponse:
    """
    Ottiene lista di tutti i prodotti
    GET /api/products
    """
    products = [doc.to_dict() for doc in inputDocuments]
    return func.HttpResponse(
        json.dumps({"products": products, "count": len(products)}),
        mimetype="application/json"
    )

# UPDATE - Aggiorna un prodotto
@app.function_name(name="UpdateProduct")
@app.route(route="products/{id}", methods=["PUT"])
@app.cosmos_db_input(
    arg_name="inputDocument",
    database_name="ProductsDB",
    container_name="Products",
    id="{id}",
    partition_key="{id}",
    connection="CosmosDBConnection"
)
@app.cosmos_db_output(
    arg_name="outputDocument",
    database_name="ProductsDB",
    container_name="Products",
    connection="CosmosDBConnection"
)
def update_product(
    req: func.HttpRequest,
    inputDocument: func.DocumentList,
    outputDocument: func.Out[func.Document]
) -> func.HttpResponse:
    """
    Aggiorna un prodotto esistente
    PUT /api/products/{id}
    Body: {"name": "New Name", "price": 149.99}
    """
    if not inputDocument:
        return func.HttpResponse("Product not found", status_code=404)
    
    try:
        updates = req.get_json()
        product = inputDocument[0].to_dict()
        
        # Aggiorna i campi
        product.update(updates)
        
        # Salva il prodotto aggiornato
        outputDocument.set(func.Document.from_dict(product))
        
        return func.HttpResponse(
            json.dumps({"message": "Product updated", "product": product}),
            mimetype="application/json"
        )
    except ValueError:
        return func.HttpResponse("Invalid JSON", status_code=400)

# DELETE - Elimina un prodotto (con notifica via queue)
@app.function_name(name="DeleteProduct")
@app.route(route="products/{id}", methods=["DELETE"])
@app.queue_output(
    arg_name="notificationQueue",
    queue_name="product-deletions",
    connection="AzureWebJobsStorage"
)
def delete_product(
    req: func.HttpRequest,
    notificationQueue: func.Out[str]
) -> func.HttpResponse:
    """
    Elimina un prodotto (implementazione semplificata)
    DELETE /api/products/{id}
    
    Invia una notifica alla coda per processing asincrono
    """
    product_id = req.route_params.get('id')
    
    # In un caso reale, elimineresti dal database qui
    # Per semplicità, inviamo solo la notifica
    
    notification = json.dumps({
        "action": "delete",
        "product_id": product_id,
        "timestamp": "2025-01-01T00:00:00Z"
    })
    
    notificationQueue.set(notification)
    
    return func.HttpResponse(
        json.dumps({"message": f"Product {product_id} deletion queued"}),
        mimetype="application/json"
    )

# Funzione di background che processa le eliminazioni
@app.function_name(name="ProcessDeletions")
@app.queue_trigger(
    arg_name="msg",
    queue_name="product-deletions",
    connection="AzureWebJobsStorage"
)
def process_deletions(msg: func.QueueMessage) -> None:
    """
    Processa le eliminazioni in modo asincrono
    Trigger: Queue message
    """
    message_data = json.loads(msg.get_body().decode('utf-8'))
    product_id = message_data.get('product_id')
    
    logging.info(f"Processing deletion for product: {product_id}")
    
    # Qui andresti effettivamente ad eliminare dal database
    # e magari inviare email di notifica, aggiornare analytics, ecc.
    
    logging.info(f"Product {product_id} successfully deleted")
````
