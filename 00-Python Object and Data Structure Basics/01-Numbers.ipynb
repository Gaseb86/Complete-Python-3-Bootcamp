### Aritmetica di base

Python può eseguire tutte le operazioni aritmetiche di base:

- Addizione
- Sottrazione

- Divisione

Ecco alcuni esempi pratici.

 
 
Now let's start with some basic arithmetic.

### Basic Arithmetic

```python
# Addition
2+1
```

```python
# Subtraction
2-1
```

```python
# Multiplication
2*2
```

```python
# Division
3/2
```

```python
# Floor Division
7//4
```

### Divisione classica vs divisione intera

La divisione classica restituisce un numero decimale, mentre la divisione intera restituisce solo la parte intera del risultato.

**So what if we just want the remainder after division?**

```python
# Modulo
7%4
```

4 goes into 7 once, with a remainder of 3. The % operator returns the remainder after division.

### Arithmetic continued

```python
# Powers
2**3
```

```python
# Can also do roots this way
4**0.5
```

```python
# Order of Operations followed in Python
2 + 10 * 10 + 3
```

```python
# Can use parentheses to specify orders
(2+10) * (10+3)
```

## Assegnazione di oggetti in Python

In Python, puoi assegnare valori a variabili usando il simbolo =.
Vediamo come funziona.

```python
# Let's create an object called "a" and assign it the number 5
a = 5
```

Now if I call *a* in my Python script, Python will treat it as the number 5.

```python
# Adding the objects
a+a
```

What happens on reassignment? Will Python let us write it over?

```python
# Reassignment
a = 10
```

```python
# Check
a
```

Yes! Python allows you to write over assigned variable names. We can also use the variables themselves when doing the reassignment. Here is an example of what I mean:

```python
# Check
a
```

```python
# Use A to redefine A
a = a + a
```

```python
# Check 
a
```
